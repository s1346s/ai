```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough
```

```python
llm = ChatOllama(model="llama3.2:1b")
parser = StrOutputParser()
```

```python
food_prompt = PromptTemplate(
    template="""
Recommend the most famous food from the country below.
Country: {country}
Return only the name of the food.
""",
    input_variables=["country"]
)
```

```python
food_chain = food_prompt | llm | parser
```

```python
food_chain.invoke({"country": "Korea"})
```

```python
recipe_prompt = PromptTemplate(
    template="""
Provide a simple recipe for the following dish:
Dish: {food}
Return the recipe in short steps.
""",
    input_variables=["food"]
)
```

```python
recipe_chain = recipe_prompt | llm | parser
```

```python
recipe_chain.invoke({"food": "Bibimbap"})
```

```python
final_chain = (
    {"country": RunnablePassthrough()}
    | {"food": food_chain}
    | recipe_chain
)

final_chain.invoke("Korea")
```
